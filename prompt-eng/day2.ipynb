````python
import os
from dotenv import load_dotenv
from openai import OpenAI
````

````python
load_dotenv(override=True)
openai_api_key=os.getenv('OPENAI_API_KEY')

client = OpenAI(api_key=openai_api_key)
````

````python
def get_response(prompt):
    response = client.chat.completions.create(
  model="gpt-4o-mini",
  messages=[{"role":"user","content":prompt}],
  max_completion_tokens=150,
  temperature=0)
    return response.choice[0].message.content
````

````python
story = "The Whispering WoodsOnce, the green valley of Oakhaven echoed with birds and the rustle of endless leaves. Maya loved the old oak at the center of the woods. Its branches touched the sky, and its roots held the rich earth in place.Then came the machines. Men with loud saws arrived, promising gold and quick wealth. They cut down the ancient trees to build roads and clear space for cattle. Maya watched the green canopy vanish day by day."


prompt = f"""Complete the story delimited by triple backticks. 
 ```{story}```"""

# Get the generated response 
response = get_response(prompt)

print("\n Original story: \n", story)
print("\n Generated story: \n", response)
````
